```python
import numpy as np
import matplotlib.pyplot as plt
```

Вектор - это одномерный массив чисел. Пример векторов.
A vector is a one-dimensional array of numbers. Examples of vectors.

```python
#Создание вектор | Create vector
v = np.array([1, 2, 3])

#Вывод вектора | Vector output
print(v)

#Вывод размерности вектора | Outputting the vector's dimensionality
print(v.shape)

#Доступ к елементам вектора | Accessing vector elements
print(v[0])
```

Основные операции с векторами включают в себя - сложение, вычетание, умножение, умножение на скаляр и скалярное произведение.
The main operations with vectors include addition, subtraction, multiplication, scalar multiplication and dot product.

```python
#Создание векторов | Creating vectors
v1 = np.array([1, 2, 3])
v2 = np.array([2, 3, 4])

#Сложение векторов | Vector addition
v_sum = v1 + v2
print("Сума сложения векторов v1 и v2 | Sum of vectors v1 and v2:", v_sum)

#Вычетание векторов | Vector subtraction
v_diff = v1 - v2
print("Cума вычетания векторов v1 и v2 | Difference of vectors v1 and v2:", v_diff)

#Умножение векторов | Vector multiplication
v_mult = v1 * v2
print("Сума умножения векторов v1 и v2 | Product of vectors v1 and v2:", v_mult)

#Умножение вектора v1 на скаляр | Scalar multiplication of vector v1
scalar = 2
v_scalar_mult = v1 * scalar
print("Сума умножения на скаляр вектора v1 | Scalar multiplication of vector v1:", v_scalar_mult)

#Скалярное произведение векторов | Dot product of vectors
v_dot = np.dot(v1, v2)
print("Скалярное произведение векторов v1 и v2 | Dot product of vectors v1 and v2:", v_dot)
```
