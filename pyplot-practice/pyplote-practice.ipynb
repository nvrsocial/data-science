```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 5]


# Функция plot() строит линейный график | The plot() function creates a line plot
plt.plot(x, y)

# Функция scatter() строит точечный график | The scatter() function creates a scatter plot
plt.scatter(x, y)

# Функция bar() строит столбчатую диаграмму | The bar() function creates a bar chart
plt.bar(x, y)

# Функция show() отображает график | The show() function displays the plot
plt.show()


# Функция pie() строит круговую диаграмму | The pie() function creates a pie chart
plt.pie(x)
plt.show()


# Функция hist() строит гистограмму | The hist() function creates a histogram
plt.hist(y)
plt.show()


# Функция title() устанавливает заголовок графика | The title() function sets the plot title
plt.plot(x, y)
plt.title("Мой график | My plot")
plt.show()


# Функция xlabel() устанавливает подпись оси X | The xlabel() function sets the X-axis label
plt.plot(x, y)
plt.xlabel("Ось X | X-axis")
plt.show()


# Функция ylabel() устанавливает подпись оси Y | The ylabel() function sets the Y-axis label
plt.plot(x, y)
plt.ylabel("Ось Y | Y-axis")
plt.show()


# Функция grid() добавляет сетку | The grid() function adds a grid
plt.plot(x, y)
plt.grid()
plt.show()


# Функция legend() отображает легенду | The legend() function displays the legend
plt.plot(x, y, label="Данные | Data")
plt.legend()
plt.show()


# Параметр color задаёт цвет линии | The color parameter sets the line color
plt.plot(x, y, color="red")
plt.show()


# Параметр linestyle задаёт стиль линии | The linestyle parameter sets the line style
plt.plot(x, y, linestyle="--")
plt.show()


# Параметр linewidth задаёт толщину линии | The linewidth parameter sets the line width
plt.plot(x, y, linewidth=3)
plt.show()


# Параметр marker задаёт маркеры точек | The marker parameter sets point markers
plt.plot(x, y, marker="o")
plt.show()


# Можно использовать несколько параметров одновременно | Multiple parameters can be used together
plt.plot(x, y, color="blue", linestyle="--", linewidth=2, marker="o")
plt.show()

# Функция xlim() задаёт диапазон оси X | The xlim() function sets the X-axis range
plt.plot(x, y)
plt.xlim(0, 6)
plt.show()


# Функция ylim() задаёт диапазон оси Y | The ylim() function sets the Y-axis range
plt.plot(x, y)
plt.ylim(0, 10)
plt.show()


# Функция xticks() изменяет метки оси X | The xticks() function changes X-axis tick labels
plt.plot(x, y)
plt.xticks([1, 2, 3, 4, 5])
plt.show()


# Функция yticks() изменяет метки оси Y | The yticks() function changes Y-axis tick labels
plt.plot(x, y)
plt.yticks([0, 2, 4, 6, 8])
plt.show()


# Функция savefig() сохраняет график в файл | The savefig() function saves the plot to a file
plt.plot(x, y)
plt.savefig("graph.png")
plt.show()


# Параметр dpi задаёт качество изображения | The dpi parameter sets the image resolution
plt.plot(x, y)
plt.savefig("graph.png", dpi=300)
plt.show()


# Можно построить несколько линий на одном графике | Multiple lines can be plotted on one figure
y2 = [1, 3, 5, 4, 6]

plt.plot(x, y, label="Первый график | First plot")
plt.plot(x, y2, label="Второй график | Second plot")

plt.legend()
plt.grid()
plt.show()

# Функция subplot() позволяет разместить несколько графиков в одном окне
# The subplot() function allows multiple plots in one figure

plt.subplot(2, 1, 1)
plt.plot(x, y)
plt.title("Линейный график | Line plot")

plt.subplot(2, 1, 2)
plt.scatter(x, y)
plt.title("Точечный график | Scatter plot")

plt.tight_layout()
plt.show()


# Функция tight_layout() автоматически регулирует расположение элементов
# The tight_layout() function automatically adjusts the layout
plt.tight_layout()


# Функция figure() создаёт отдельное окно для графика
# The figure() function creates a separate figure
plt.figure()

plt.plot(x, y)
plt.show()


# Функция close() закрывает текущую фигуру
# The close() function closes the current figure
plt.close()

# Функция annotate() добавляет текстовую аннотацию на график
# The annotate() function adds a text annotation to the plot
plt.plot(x, y)

plt.annotate("Максимум | Maximum", xy=(4, 7), xytext=(3, 8))

plt.show()
```
